# Handling Missing Values with pandas

A compact preprocessing notebook using Seaborn's Titanic dataset to explore:
- missing-value counts;
- row/column deletion;
- mean imputation;
- median imputation;
- mode imputation for categorical data.

This is a learning notebook focused on preprocessing choices rather than model training.

```python
import seaborn as sns
df = sns.load_dataset("titanic")
df.head()
```

```python
# checking for missing values
df.isnull().sum()
```

```python
df.shape
```

```python
df.dropna().shape
```

```python
#column wise deletion
df.dropna(axis=1)
```

```python
## MEAN VALUE IMPUTATION
sns.histplot(df['age'],kde =True)
```

```python
df['age_mean'] = df['age'].fillna(df['age'].mean())
df[['age_mean',"age"]]
```

```python
df['age'].describe()
```

```python
sns.boxplot(x=df['age'])
```

```python
df['age'].isnull().mean() * 100
```

```python
df['age'].mean(),df['age'].median()
```

```python
## median value imputation : if we have outliers in the dataset
```

```python
df['age_median']=df['age'].fillna(df['age'].median())
```

```python
df[['age','age_median','age_mean']]
```

```python
## mode imputation Technique(categorical values)
```

```python
df[df['embarked'].isnull()]
```

```python
df['embarked'].unique()
```

```python
mode_value = df[df['embarked'].notna()]['embarked'].mode()[0]
df['embarked_mode'] = df['embarked'].fillna(mode_value)
```

```python
df[['embarked','embarked_mode']]
```

```python
df['embarked_mode'].isnull().sum()
```

```python
df['embarked'].isnull().sum()
```
