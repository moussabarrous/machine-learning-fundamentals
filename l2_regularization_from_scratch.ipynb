# L2 Regularization from Scratch

This notebook implements the core equations for **L2-regularized linear and logistic regression** with NumPy.

It focuses on:
- regularized cost functions;
- regularized gradients;
- a reusable gradient-descent function.

Small random arrays are used as sanity checks so the notebook remains self-contained.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def compute_cost_linear_reg(X,y,w,b,lambda_=1):
  m = X.shape[0]
  n = len(w)
  cost = 0
  for i in range(m):
    f_wb = np.dot(X[i],w)+b
    err = f_wb - y[i]
    cost = cost + err**2
  cost = cost/(2*m)

  reg_cost = 0
  for j in range(n):
    reg_cost += (w[j]**2)
  reg_cost = (lambda_ / (2*m)) * reg_cost

  total_cost = cost + reg_cost
  return total_cost
```

```python
np.random.seed(1)
X_tmp = np.random.rand(5,6)
y_tmp = np.array([0,1,0,1,0])
w_tmp = np.random.rand(X_tmp.shape[1]).reshape(-1,)-0.5
b_tmp = 0.5
lambda_tmp = 0.7
cost_tmp = compute_cost_linear_reg(X_tmp, y_tmp, w_tmp, b_tmp, lambda_tmp)

print("Regularized cost:", cost_tmp)
```

```python
def sigmoid(z_i):
  return 1/(1+np.exp(-z_i))
```

```python
def compute_cost_logistic_reg(X, y, w, b, lambda_ = 1):
  m = X.shape[0]
  n = X.shape[1]
  cost = 0
  for i in range(m):
    z_i = np.dot(X[i],w)+b
    f_wb_i = sigmoid(z_i)
    cost+= -y[i]*np.log(f_wb_i) - (1-y[i])*np.log(1-f_wb_i)
  cost = cost/m

  reg_cost = 0
  for j in range(n):
    reg_cost+= (w[j]**2)
  reg_cost = (lambda_ /(2*m))*reg_cost
  total_cost = cost + reg_cost
  return total_cost
```

```python
np.random.seed(1)
X_tmp = np.random.rand(5,6)
y_tmp = np.array([0,1,0,1,0])
w_tmp = np.random.rand(X_tmp.shape[1]).reshape(-1,)-0.5
b_tmp = 0.5
lambda_tmp = 0.7
cost_tmp = compute_cost_logistic_reg(X_tmp, y_tmp, w_tmp, b_tmp, lambda_tmp)

print("Regularized cost:", cost_tmp)
```

```python
def compute_gradient_linear_reg(X,y,w,b,lambda_):
  m = X.shape[0]
  n = X.shape[1]
  dj_dw = np.zeros(n)
  dj_db = 0
  for i in range(m):
    f_wb_i = np.dot(X[i],w)+b
    err = f_wb_i - y[i]
    for j in range(n):
      dj_dw[j] += err * X[i,j]
    dj_db += err

  dj_dw = dj_dw / m
  dj_db = dj_db / m
  for j in range(n):
    dj_dw[j] = dj_dw[j]+(lambda_ / m )* (w[j])
  return dj_dw , dj_db

```

```python
np.random.seed(1)
X_tmp = np.random.rand(5,3)
y_tmp = np.array([0,1,0,1,0])
w_tmp = np.random.rand(X_tmp.shape[1])
b_tmp = 0.5
lambda_tmp = 0.7
dj_dw_tmp, dj_db_tmp =  compute_gradient_linear_reg(X_tmp, y_tmp, w_tmp, b_tmp, lambda_tmp)

print(f"dj_db: {dj_db_tmp}", )
print(f"Regularized dj_dw:\n {dj_dw_tmp.tolist()}", )
```

```python
def compute_gradient_logistic_reg(X, y, w, b, lambda_):

 m = X.shape[0]
 n = X.shape[1]
 dj_dw = np.zeros(n)
 dj_db = 0
 for i in range(m):
   z_i = np.dot(X[i],w)+b
   f_wb_i = sigmoid(z_i)
   err_i = f_wb_i - y[i]
   for j in range(n):
     dj_dw[j]+= err_i*X[i,j]
   dj_db +=err_i
 dj_dw = dj_dw / m
 dj_db = dj_db / m

 for j in range(n):
   dj_dw[j]+= (lambda_ / m) * w[j]

 return dj_dw , dj_db
```

```python
np.random.seed(1)
X_tmp = np.random.rand(5,3)
y_tmp = np.array([0,1,0,1,0])
w_tmp = np.random.rand(X_tmp.shape[1])
b_tmp = 0.5
lambda_tmp = 0.7
dj_dw_tmp, dj_db_tmp =  compute_gradient_logistic_reg(X_tmp, y_tmp, w_tmp, b_tmp, lambda_tmp)

print(f"dj_db: {dj_db_tmp}", )
print(f"Regularized dj_dw:\n {dj_dw_tmp.tolist()}", )
```

```python
def gradient_descent(X, y, w_in, b_in, alpha, num_iters, lambda_):
    w = w_in.copy()
    b = b_in

    J_history = []

    for i in range(num_iters):
        dj_dw, dj_db = compute_gradient_logistic_reg(X, y, w, b, lambda_)

        # update parameters
        w = w - alpha * dj_dw
        b = b - alpha * dj_db

        # save cost
        cost = compute_cost_logistic_reg(X, y, w, b, lambda_)
        J_history.append(cost)

        # print sometimes
        if i % max(1, num_iters // 10) == 0:
            print(f"Iteration {i}: Cost {cost}")

    return w, b, J_history
```
