# Logistic Regression from Scratch

A NumPy implementation of binary logistic regression including the sigmoid function, log-loss cost, analytical gradients, gradient descent, classification, accuracy calculation, and visualization.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
X_train = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y_train = np.array([0, 0, 0, 1, 1, 1])
```

```python
def sigmoid(z):
  return 1/(1+np.exp(-z))
```

```python
def cost(x,y,w,b):
  m = x.shape[0]
  z = np.dot(x,w)+b
  f_wb = sigmoid(z)
  cost = -y*np.log(f_wb)-(1-y)*np.log(1-f_wb)
  cost = (1/m)*(np.sum(cost))
  return cost
```

```python
def gradient(x,y,w,b):
  m = x.shape[0]
  n = x.shape[1]
  dj_dw = np.zeros(n)
  dj_db = 0
  for i in range(m):
    z = np.dot(x[i],w)+b
    f_wb_i = sigmoid(z)
    err_i= f_wb_i - y[i]
    for j in range(n):
      dj_dw[j] += err_i * x[i,j]
    dj_db +=err_i

  dj_dw = dj_dw / m
  dj_db = dj_db / m
  return dj_dw , dj_db
```

```python
def gradient_decsent(x,y,w,b,iter,alpha):
  j_history = []
  for i in range(iter):
    dj_dw , dj_db = gradient(x,y,w,b)
    w = w - alpha*dj_dw
    b = b -alpha*dj_db
    current_cost = cost(x,y,w,b)
    j_history.append(cost(x,y,w,b))
    if i % 1000 == 0:
      print(f"iteration: {i}, cost: {current_cost:.6f}")
  return w,b,j_history
```

```python
w_init = np.zeros(X_train.shape[1])  # 2 weights because X_train has 2 columns
b_init = 0

alpha = 0.1
iter = 10000
w_final ,b_final ,j_history= gradient_decsent(X_train,y_train,w_init,b_init,iter,alpha)
print(f"w_final est : {w_final}")
print(f"b final est : {b_final}")
```

```python
def predict(x, w, b):
    prob = sigmoid(np.dot(x, w) + b)
    return (prob >= 0.5).astype(int)

# verify on training data
pred = predict(X_train, w_final, b_final)
acc = np.mean(pred == y_train) * 100

print("predictions:", pred)
print("real labels:", y_train)
print("accuracy:", acc, "%")
```

```python
pred = predict(X_train, w_final, b_final)

plt.scatter(X_train[:,0], X_train[:,1], c=y_train, s=200, alpha=0.5, label="real")
plt.scatter(X_train[:,0], X_train[:,1], marker="x", c=pred, s=120, linewidths=3, label="pred")

plt.xlabel("x0")
plt.ylabel("x1")
plt.legend()
plt.show()
```
