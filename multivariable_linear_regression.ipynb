# Multivariable Linear Regression from Scratch

A compact NumPy experiment implementing normalization, cost, gradient computation, gradient descent, and predictions for a multivariable linear-regression problem.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
X_train = np.array([[2104,5,1,45],[1416,3,2,40],[852,2,1,35]])
y_train = np.array([460,232,178])
```

```python
X_train.shape
y_train.shape
```

```python
def normalize(x):
  mean = np.mean(x,axis=0)
  std = np.std(x,axis=0)
  x_n = (x - mean)/std
  return x_n
```

```python
x_n = normalize(X_train)
print(x_n)
```

```python
w_int = np.zeros(X_train.shape[1])
b_int = 0
```

```python
def cost (x,y,w,b):
  m = x.shape[0]
  f_wb = np.dot(x,w)+b
  err = f_wb - y
  J = np.sum(err**2)*(1/(2*m))
  return J
```

```python
## Computing gradient
def gradient(x,y,w,b):
  m = x.shape[0]
  n = x.shape[1]
  dj_dw = np.zeros(n)
  f_wb = np.dot(x,w)+b
  err = f_wb - y
  for i in range (n):
    dj_dw[i] = np.sum(err*x[:,i]) * (1/m)
  dj_db = np.sum(err) *(1/m)
  return dj_dw , dj_db
```

```python
def gradient_descent(x,y,w,b,alpha,iter,gradient,cost):
  j_history = []
  for i in range(iter):
     dj_dw ,dj_db = gradient(x,y,w,b)
     w = w - alpha*dj_dw
     b = b - alpha*dj_db
     j_history.append(cost(x,y,w,b))
  return w ,b ,j_history
```

```python
w_final , b_final , j_history = gradient_descent(x_n , y_train , w_int , b_int ,0.01,10000,gradient,cost)
print(f"The final w is : {w_final}")
print(f"the finaal b is :{b_final}")
```

```python
plt.plot(j_history)
plt.xlabel("iterations")
plt.ylabel("cost")
```

```python
plt.plot(j_history[10:])
plt.xlabel("iterations")
plt.ylabel("cost")
plt.title("Cost after first 10 iterations")
plt.show()
```

```python
## Predictions
y_predict = x_n@w_final + b_final
print(y_predict)
```

```python
plt.scatter(X_train[:,0],y_train,label="target")
plt.scatter(X_train[:,0],y_predict,label ="predictions")
plt.xlabel("floors")
plt.ylabel("prices")
plt.legend()
plt.show()
```

```python
plt.scatter(X_train[:,0], y_train, label="target", marker="o")
plt.scatter(X_train[:,0], y_predict, label="predictions", marker="x")
plt.xlabel("floors")
plt.ylabel("prices")
plt.legend()
plt.show()
```
