# Building a Linear Regression Model from Scratch

A self-contained NumPy implementation of multivariable linear regression, including feature normalization, the cost function, analytical gradients, gradient descent, predictions, and error analysis.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python


X_raw = np.array([
    [1200,   1,   50000],
    [2500,   2,   120000],
    [4000,   1,   200000],
    [5200,   3,   350000],
    [8000,   2,   600000],
    [10000,  4,   800000],
    [15000,  3,   1200000],
    [18000,  5,   1500000],
    [22000,  4,   1900000],
    [26000,  6,   2400000],
    [31000,  5,   3000000],
    [36000,  7,   3500000],
    [42000,  6,   4200000],
    [50000,  8,   5000000],
    [58000,  7,   5800000],
    [65000,  9,   6700000],
    [72000, 10,   7500000],
    [80000,  8,   8500000],
    [90000, 11,   9800000],
    [100000,12,   11000000],
    [115000,10,   12500000],
    [130000,13,   14000000],
    [145000,14,   15800000],
    [160000,12,   17600000],
    [180000,15,   20000000],
    [200000,16,   22500000],
    [225000,14,   25000000],
    [250000,18,   28000000],
    [280000,17,   31500000],
    [320000,20,   36000000]
], dtype=float)

y = np.array([
    40.75, 44.05, 51.00, 60.25, 62.00,
    90.00, 94.50, 122.50, 125.50, 157.00,
    166.50, 203.50, 223.00, 259.00, 294.00,
    328.00, 368.50, 384.50, 448.00, 491.00,
    550.00, 613.00, 688.50, 739.00, 842.00,
    930.50, 1027.50, 1142.00, 1271.50, 1450.00
])
```

```python
X_raw.shape
y.shape
```

```python
def z_normalization(X):
  mean = np.mean(X,axis=0)
  sigma = np.std(X,axis=0)
  X_n = (X - mean)/sigma
  return X_n
```

```python
X_n = z_normalization(X_raw)
print(X_n)
```

```python
w_int = np.zeros(X_n.shape[1])
b_int = 0
```

```python
def cost(x,y,w,b):
  m = x.shape[0]
  f_wb = np.dot(x , w) + b
  err = f_wb - y
  cost = (1/(2*m) )*np.sum(err**2)
  return cost
```

```python
def gradient(x,y,w,b):
  dj_dw = np.zeros(x.shape[1])
  dj_db = 0
  m =x.shape[0]
  n = x.shape[1]
  f_wb =np.dot(x,w)+b
  err = f_wb - y
  for j in range (n):
    dj_dw[j] = np.sum(err * x[:,j])*(1/m)
  dj_db = np.sum(err)*(1/m)
  return dj_dw , dj_db
```

```python
def gradient_descent(x,y,w,b,alpha,iteration,gradient,cost):
  j_history = []
  for i in range(iteration):
    dj_dw , dj_db = gradient(x,y,w,b)
    w = w -(alpha*dj_dw)
    b = b - (alpha*dj_db)
    j_history.append(cost(x,y,w,b))
  final_w = w
  final_b = b
  return final_w,final_b,j_history
```

```python
final_w , final_b , j_history = gradient_descent(X_n , y , w_int , b_int , alpha =0.01,iteration =50000 ,gradient=gradient , cost=cost)
print(f"final w is : {final_w}")
print(f"final b is : {final_b}")
final_w = final_w
final_b = final_b
```

```python
plt.plot(j_history)
plt.xlabel('iterations')
plt.ylabel("j")
plt.title("Cost over iterations")
plt.show()
```

```python
y_pred = X_n @ final_w + final_b
```

```python
print(y_pred)
```

```python
plt.scatter(y, y_pred)
plt.xlabel("Actual y")
plt.ylabel("Predicted y")
plt.title("Actual vs Predicted")
plt.show()
```

```python
plt.scatter(range(len(y)), y, label="Actual y")
plt.scatter(range(len(y_pred)), y_pred, label="Predicted y")

plt.xlabel("Example number")
plt.ylabel("y value")
plt.title("Actual vs Predicted Values")
plt.legend()
plt.show()
```

```python
errors = y_pred - y

print("Max error:", np.max(np.abs(errors)))
print("Mean absolute error:", np.mean(np.abs(errors)))
print("Final cost:", j_history[-1])
```
